AUROC curve

```python

sofa_df = pd.read_csv("../sepsis_sofa.csv")

y_true_sofa = sofa_df["sepsis"].values
y_score_sofa = sofa_df["sepsis_sofa"].values

fpr, tpr, _ = roc_curve(y_true_sofa, y_score_sofa)
auc_val = roc_auc_score(y_true_sofa, y_score_sofa)

roc_data.append({
    "Model": "SOFA",
    "Fold": "All",
    "fpr": fpr,
    "tpr": tpr,
    "auc": auc_val
})

precision, recall, _ = precision_recall_curve(y_true_sofa, y_score_sofa)
ap_val = average_precision_score(y_true_sofa, y_score_sofa)

prc_data.append({
    "Model": "SOFA",
    "Fold": "All",
    "precision": precision,
    "recall": recall,
    "ap": ap_val
})

def plot_mean_roc_curves(roc_data):

    if not roc_data:
        print("No ROC data available for plotting")
        return
        
    plt.figure(figsize=(8, 8))
    
    base_fpr = np.linspace(0, 1, 100)
    models = list(dict.fromkeys(d["Model"] for d in roc_data))
    colors = ["#1f77b4","#d62728","#2ca02c","#9467bd","#8c564b","#e377c2","#17becf"]
    
    for i, model_name in enumerate(models):
        if model_name == "SOFA":
            data = [d for d in roc_data if d["Model"] == "SOFA"][0]
            plt.plot(
                data["fpr"], data["tpr"],
                color="black",
                linestyle="--",
                linewidth=2.5,
                label=f"SOFA (AUC = {data['auc']:.3f})"
            )
            continue
        
        model_data = [data for data in roc_data if data['Model'] == model_name]
        
        tprs = []
        aucs = []
        
        for fold_data in model_data:
            fpr = fold_data['fpr']
            tpr = fold_data['tpr']

            interp_tpr = np.interp(base_fpr, fpr, tpr)
            interp_tpr[0] = 0.0
            tprs.append(interp_tpr)
            aucs.append(fold_data['auc'])
        
        mean_tpr = np.mean(tprs, axis=0)
        mean_tpr[-1] = 1.0
        mean_auc = np.mean(aucs)
        std_auc = np.std(aucs)
        
        plt.plot(base_fpr, mean_tpr, color=colors[i], 
                label=f'{model_name} (AUC = {mean_auc:.3f} ± {std_auc:.3f})',
                linewidth=2)
    
    plt.plot([0, 1], [0, 1], 'k--', alpha=0.5, label='Random Classifier')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate', fontsize=14)
    plt.ylabel('True Positive Rate', fontsize=14)
    plt.title('Receiver Operating Characteristic (ROC) Curves', fontsize=16)
    plt.legend(loc='lower right', fontsize=14)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    if savefigs == 0:
        plt.show()
    else:
        plt.savefig("../roc_first.svg", dpi=300, bbox_inches="tight")
    plt.close()

plot_mean_roc_curves(roc_data)

def plot_mean_prc_curves(prc_data):

    if not prc_data:
        print("No PRC data available for plotting")
        return

    plt.figure(figsize=(8, 8))

    base_recall = np.linspace(0, 1, 100)
    models = list(dict.fromkeys(data["Model"] for data in prc_data))
    colors = ["#1f77b4","#d62728","#2ca02c","#9467bd","#8c564b","#e377c2","#17becf"]

    for i, model_name in enumerate(models):
        if model_name == "SOFA":
            data = [d for d in prc_data if d["Model"] == "SOFA"][0]
            plt.plot(
                data["recall"], data["precision"],
                color="black",
                linestyle="--",
                linewidth=2.5,
                label=f"SOFA (AP = {data['ap']:.3f})"
            )
            continue
            
        model_data = [d for d in prc_data if d['Model'] == model_name]

        precisions = []
        aps = []

        for fold_data in model_data:
            recall = fold_data['recall']
            precision = fold_data['precision']

            interp_precision = np.interp(
                base_recall,
                recall[::-1],
                precision[::-1]
            )

            precisions.append(interp_precision)
            aps.append(fold_data['ap'])

        mean_precision = np.mean(precisions, axis=0)

        mean_ap = np.mean(aps)
        std_ap = np.std(aps)

        plt.plot(
            base_recall,
            mean_precision,
            color=colors[i],
            linewidth=2,
            label=f'{model_name} (AP = {mean_ap:.3f} ± {std_ap:.3f})'
        )

    all_y = []
    for d in prc_data:
        all_y.append(d['ap'])

    plt.xlabel('Recall', fontsize=14)
    plt.ylabel('Precision', fontsize=14)
    plt.title('Precision–Recall Curves (PRC)', fontsize=16)
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.legend(loc='lower left', fontsize=14)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    if savefigs == 0:
        plt.show()
    else:
        plt.savefig("../prc_first.svg", dpi=300, bbox_inches="tight")
    plt.close()

plot_mean_prc_curves(prc_data)
```

SHAP

```python

import shap
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import xgboost as xgb
shap.initjs()

shap_fold_results = []
feature_names = None

for fold, model in enumerate(xgbmodels, start=1):
    print(f"SHAP for fold {fold}")

    test_path = f"{data_prefix}_test_{fold}_processed.csv"
    test_df = pd.read_csv(test_path)

    y_test = test_df["sepsis"]
    X_test = test_df.drop(columns=["sepsis"])
    if fold == 3:
        X_test_fold3 = X_test

    if feature_names is None:
        feature_names = X_test.columns.tolist()
        
    X_test_shap = X_test.sample(
        n=min(1000, len(X_test)),
        random_state=42
    )
    explainer = shap.Explainer(model, X_test_shap)
    shap_values = explainer(X_test_shap).values

    # binary classification: shap_values shape = (n_samples, n_features)
    mean_abs_shap = np.mean(np.abs(shap_values), axis=0)

    shap_fold_results.append(mean_abs_shap)

shap_array = np.vstack(shap_fold_results)   # (n_folds, n_features)

mean_shap = shap_array.mean(axis=0)
std_shap  = shap_array.std(axis=0)

shap_stats = pd.DataFrame({
    "Feature": feature_names,
    "Mean_SHAP": mean_shap,
    "STD_SHAP": std_shap
}).sort_values("Mean_SHAP", ascending=False)

plt.figure(figsize=(7, 5))

top20 = shap_stats.head(10)

plt.barh(
    top20["Feature"][::-1],
    top20["Mean_SHAP"][::-1],
    xerr=top20["STD_SHAP"][::-1],
    capsize=4
)

plt.yticks(fontsize=12)
plt.xlabel("Mean |SHAP value|", fontsize=14)
plt.title("Top 10 Features by Mean SHAP Value (with Std)", fontsize=14)
plt.tight_layout()
if savefigs == 0:
    plt.show()
else:
    plt.savefig("../shapmean_first.svg", dpi=300)
plt.close()

X_test_fold3_sub = X_test_fold3.sample(n=1000, random_state=42)

booster = xgbmodels[2].get_booster()
shap_values_fold3 = booster.predict(
    xgb.DMatrix(X_test_fold3_sub),
    pred_contribs=True
)[:, :-1]

top5_features = shap_stats["Feature"].head(10).tolist()
top5_idx = [X_test_fold3_sub.columns.get_loc(f) for f in top5_features]
shap_values_top5 = shap_values_fold3[:, top5_idx]
X_test_top5 = X_test_fold3_sub[top5_features]

shap.summary_plot(shap_values_top5, X_test_top5, plot_type="dot", show=False)
if savefigs == 0:
    plt.show()
else:
    plt.savefig("../shap_first.svg", dpi=300)
plt.close()
```

DCA

```python
def compute_net_benefit(y_true, y_score, thresholds):
    N = len(y_true)
    net_benefits = []

    for pt in thresholds:
        y_pred = (y_score >= pt).astype(int)

        TP = np.sum((y_pred == 1) & (y_true == 1))
        FP = np.sum((y_pred == 1) & (y_true == 0))

        nb = (TP / N) - (FP / N) * (pt / (1 - pt))
        net_benefits.append(nb)

    return np.array(net_benefits)


def plot_dca(dca_data):
    thresholds = np.linspace(0.01, 0.99, 99)

    colors = [
        "#1f77b4",
        "#d62728",
        "#2ca02c",
        "#9467bd",
        "#8c564b",
        "#e377c2",
        "#17becf",
    ]

    plt.figure(figsize=(8, 8))

    models = list(dict.fromkeys(d["Model"] for d in dca_data))

    # Plot model curves
    for i, model in enumerate(models):
        model_data = [d for d in dca_data if d["Model"] == model]
        nbs = []

        for d in model_data:
            nb = compute_net_benefit(d["y_true"], d["y_score"], thresholds)
            nbs.append(nb)

        mean_nb = np.mean(nbs, axis=0)

        plt.plot(
            thresholds,
            mean_nb,
            label=model,
            linewidth=2,
            color=colors[i]
        )

    # Treat-all curve
    prevalence = np.mean(
        np.concatenate([d["y_true"] for d in dca_data])
    )
    treat_all = prevalence - (1 - prevalence) * thresholds / (1 - thresholds)

    plt.plot(
        thresholds,
        treat_all,
        linestyle="--",
        color="black",
        linewidth=2,
        label="Treat All"
    )

    # Treat-none curve
    plt.plot(
        thresholds,
        np.zeros_like(thresholds),
        linestyle=":",
        color="gray",
        linewidth=2,
        label="Treat None"
    )

    plt.xlabel("Threshold probability", fontsize=14)
    plt.ylabel("Net benefit", fontsize=14)
    plt.title("Decision Curves", fontsize=16)
    plt.xlim(0, 1)
    plt.ylim(-0.1, 0.5)
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    if savefigs == 0:
        plt.show()
    else:
        plt.savefig("../dca_first.svg", dpi=300)
    plt.close()


plot_dca(dca_data)
```

Calibration

```python
from sklearn.calibration import calibration_curve

def compute_calibration(y_true, y_score, n_bins=10):
    prob_true, prob_pred = calibration_curve(
        y_true,
        y_score,
        n_bins=n_bins,
        strategy="quantile"
    )
    return prob_pred, prob_true

def plot_mean_calibration_curves(dca_data, n_bins=10):
    base_prob = np.linspace(0.0, 1.0, n_bins)
    plt.figure(figsize=(8, 8))

    models = list(dict.fromkeys(d["Model"] for d in dca_data))
    colors = ["#1f77b4","#d62728","#2ca02c","#9467bd","#8c564b","#e377c2","#17becf"]

    for i, model in enumerate(models):
        model_data = [d for d in dca_data if d["Model"] == model]

        preds_all = []
        trues_all = []

        for d in model_data:
            prob_pred, prob_true = compute_calibration(
                d["y_true"], d["y_score"], n_bins
            )
            interp_true = np.interp(
                base_prob,
                prob_pred,
                prob_true,
                left=prob_true[0],
                right=prob_true[-1]
            )
            
            preds_all.append(base_prob)
            trues_all.append(interp_true)

        # 对 bin 对齐后取均值
        mean_pred = np.mean(preds_all, axis=0)
        mean_true = np.mean(trues_all, axis=0)
        std_true = np.std(trues_all, axis=0)

        plt.plot(
            mean_pred,
            mean_true,
            marker="o",
            linewidth=2,
            color=colors[i],
            label=model
        )

        #plt.fill_between(
        #    mean_pred,
        #    mean_true - std_true,
        #    mean_true + std_true,
        #    color=colors[i % len(colors)],
        #    alpha=0.15
        #)

    # Perfect calibration
    plt.plot([0, 1], [0, 1], "k--", label="Perfect calibration")

    plt.xlabel("Mean predicted probability", fontsize=14)
    plt.ylabel("Observed event rate", fontsize=14)
    plt.title("Mean Calibration Curves", fontsize=16)
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    if savefigs == 0:
        plt.show()
    else:
        plt.savefig("../calib_first.svg", dpi=300)
    plt.close()

plot_mean_calibration_curves(dca_data, n_bins=10)

playsound('1.wav')
```
